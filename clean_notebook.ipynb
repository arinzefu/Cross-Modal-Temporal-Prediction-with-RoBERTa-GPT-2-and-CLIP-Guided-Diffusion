```python
import nbformat

notebook_path = "02_visual_autoencoder_pretraining.ipynb"

# Load notebook
with open(notebook_path, "r", encoding="utf-8") as f:
    nb = nbformat.read(f, as_version=4)

# Remove notebook-level metadata except kernelspec and language info
nb.metadata = {
    key: value
    for key, value in nb.metadata.items()
    if key in ["kernelspec", "language_info"]
}

# Remove cell metadata
for cell in nb.cells:
    cell.metadata = {}

# Save notebook
with open(notebook_path, "w", encoding="utf-8") as f:
    nbformat.write(nb, f)

print("Notebook cleaned while keeping outputs.")
```
